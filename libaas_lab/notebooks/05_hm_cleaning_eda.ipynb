```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python
file_path = '/kaggle/input/competitions/h-and-m-personalized-fashion-recommendations/articles.csv'

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nDataset information:")
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values[missing_values > 0])

print("\nTotal missing values:", missing_values.sum())

print("\nMissing percentage:")
display(
    (df.isnull().mean() * 100).round(2)
)
```

```python
print("Duplicate rows before cleaning:", df.duplicated().sum())

print("Duplicate article IDs:", df['article_id'].duplicated().sum())

print("Duplicate product codes:", df['product_code'].duplicated().sum())

# Remove only completely identical rows
df = df.drop_duplicates().copy()

print("\nShape after removing exact duplicates:", df.shape)
print("Remaining exact duplicates:", df.duplicated().sum())
```

```python
df['detail_desc'] = (
    df['detail_desc']
    .fillna('')
    .astype(str)
    .str.strip()
)

# Clean whitespace in text columns
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Missing values after cleaning:")
display(df.isnull().sum())

print("Total missing values:", df.isnull().sum().sum())
```

```python
columns_to_check = [
    'product_type_name',
    'product_group_name',
    'colour_group_name',
    'department_name',
    'perceived_colour_value_name',
    'garment_group_name'
]

for col in columns_to_check:
    print("\n", col)
    print("Unique categories:", df[col].nunique())
    print(df[col].value_counts().head(10))
```

```python
counts = df['product_type_name'].value_counts().head(15)

plt.figure(figsize=(12, 6))
sns.barplot(x=counts.values, y=counts.index)

plt.title('Top 15 Product Types')
plt.xlabel('Number of Products')
plt.ylabel('Product Type')
plt.tight_layout()
plt.show()
```

```python
counts = df['product_group_name'].value_counts()

plt.figure(figsize=(12, 6))
sns.barplot(x=counts.values, y=counts.index)

plt.title('Product Group Distribution')
plt.xlabel('Number of Products')
plt.ylabel('Product Group')
plt.tight_layout()
plt.show()
```

```python
counts = df['colour_group_name'].value_counts().head(15)

plt.figure(figsize=(12, 6))
sns.barplot(x=counts.values, y=counts.index)

plt.title('Top 15 Product Colours')
plt.xlabel('Number of Products')
plt.ylabel('Colour')
plt.tight_layout()
plt.show()
```

```python
counts = df['department_name'].value_counts().head(15)

plt.figure(figsize=(12, 6))
sns.barplot(x=counts.values, y=counts.index)

plt.title('Top 15 Fashion Departments')
plt.xlabel('Number of Products')
plt.ylabel('Department')
plt.tight_layout()
plt.show()
```

```python
df['description_length'] = df['detail_desc'].str.len()

plt.figure(figsize=(10, 5))

sns.histplot(
    df['description_length'],
    bins=40,
    kde=True
)

plt.title('Product Description Length Distribution')
plt.xlabel('Description Length')
plt.ylabel('Number of Products')

plt.tight_layout()
plt.show()

print(df['description_length'].describe())
```

```python
Q1 = df['description_length'].quantile(0.25)
Q3 = df['description_length'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['description_length'] < lower) |
    (df['description_length'] > upper)
]

print("Lower bound:", lower)
print("Upper bound:", upper)
print("Potential outliers:", len(outliers))

plt.figure(figsize=(10, 4))
sns.boxplot(x=df['description_length'])
plt.title('Description Length Outlier Analysis')
plt.show()
```

```python
import os

output_dir = '/kaggle/working/hm_phase1_output'

os.makedirs(output_dir, exist_ok=True)

df.to_csv(
    os.path.join(output_dir, 'hm_articles_cleaned.csv'),
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
import os
import zipfile
import matplotlib.pyplot as plt
import seaborn as sns

# Create folder for graphs
image_dir = "/kaggle/working/HM_EDA_Graphs"
os.makedirs(image_dir, exist_ok=True)

# Define the 4 graphs
graphs = {
    "product_types.png": (
        "product_type_name",
        "Top 15 Product Types",
        15
    ),
    "product_groups.png": (
        "product_group_name",
        "Product Group Distribution",
        15
    ),
    "product_colours.png": (
        "colour_group_name",
        "Top 15 Product Colours",
        15
    ),
    "departments.png": (
        "department_name",
        "Top 15 Fashion Departments",
        15
    )
}

# Generate and save each graph
for filename, (column, title, top_n) in graphs.items():

    counts = df[column].value_counts().head(top_n)

    plt.figure(figsize=(12, 6))

    sns.barplot(
        x=counts.values,
        y=counts.index
    )

    plt.title(title, fontsize=16)
    plt.xlabel("Number of Products")
    plt.ylabel(column.replace("_", " ").title())

    plt.tight_layout()

    save_path = os.path.join(image_dir, filename)

    plt.savefig(
        save_path,
        dpi=150,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print("Saved:", filename)


# Create ZIP file
zip_path = "/kaggle/working/HM_EDA_Graphs.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zipf:

    for filename in os.listdir(image_dir):

        file_path = os.path.join(image_dir, filename)

        zipf.write(
            file_path,
            arcname=filename
        )

print("\nAll 4 graphs saved successfully!")
print("ZIP location:", zip_path)
```
