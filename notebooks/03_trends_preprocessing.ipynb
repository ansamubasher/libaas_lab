# Step 3: Preprocessing, Normalization & Scaling

**Dataset:** `data/processed/trends_cleaned.csv`  
**Goal:** Scale numeric columns, convert categories and True/False flags into numbers, compare data before vs. after scaling, and save `data/processed/trends_processed.csv`.


## 1. Load Cleaned Data


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder

df = pd.read_csv("../data/processed/trends_cleaned.csv")
print("Loaded dataset shape:", df.shape)
df.head()
```

## 2. Why This Dataset Needs Scaling

- `year` is between `2001` and `2020`, while `rank` is between `1` and `5`.
- `query_len` (`1` to `158`) and `query_word_count` (`1` to `22`) are right-skewed with long-query outliers.
- Scaling puts these columns on similar ranges so large numbers like `year` do not overpower small numbers like `rank` in machine learning models.
- `rank` is left unscaled on purpose: it is already small (1-5), it is ordinal, and it could be a target for modeling.


## 3. Scale Numeric Columns


```python
df_proc = df.copy()

# Min-Max scaling (puts values between 0 and 1)
mm_scaler = MinMaxScaler()
# `rank` (1-5) is NOT scaled: it is an ordering inside each top-5 list and a possible target, not a feature
df_proc[["year_scaled"]] = mm_scaler.fit_transform(df_proc[["year"]])

# Log transform + Standard scaling for skewed length columns
df_proc["query_len_log"] = np.log1p(df_proc["query_len"])
df_proc["query_words_log"] = np.log1p(df_proc["query_word_count"])

std_scaler = StandardScaler()
df_proc[["query_len_scaled", "query_words_scaled"]] = std_scaler.fit_transform(
    df_proc[["query_len_log", "query_words_log"]]
)

df_proc[["year", "year_scaled", "rank", "query_len", "query_len_scaled"]].describe().round(2)
```

## 4. Before vs. After Scaling Graph


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].hist(df_proc["query_len"], bins=35, color="coral", edgecolor="white")
axes[0].set_title("Before Scaling: query_len")
axes[0].set_xlabel("Characters")
axes[0].set_ylabel("Count")

axes[1].hist(df_proc["query_len_scaled"], bins=35, color="steelblue", edgecolor="white")
axes[1].set_title("After log1p + StandardScaler: query_len_scaled")
axes[1].set_xlabel("Scaled Value")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.savefig("../images/trends/07_before_after_scaling.png", dpi=150)
plt.show()
```

**What the graph shows:** before scaling, `query_len` is right-skewed with a long tail. After `log1p` + `StandardScaler` the distribution is much closer to bell-shaped and centred on 0, so the long queries no longer dominate distance-based or linear models.


## 5. Convert Categories & True/False Columns to Numbers


```python
# location (83 values) and script (few values): label encoding is fine for tree models
for col, new_col in [("location", "location_enc"), ("script", "script_enc")]:
    encoder = LabelEncoder()
    df_proc[new_col] = encoder.fit_transform(df_proc[col].astype(str))

# category_clean has ~2,000+ unique values, so plain integer codes would be meaningless.
# Frequency encoding: replace each category with how often it appears (as a share of all rows).
cat_freq = df_proc["category_clean"].value_counts(normalize=True)
df_proc["category_freq"] = df_proc["category_clean"].map(cat_freq)
print("Unique categories:", df_proc["category_clean"].nunique())

# Convert True/False columns to 1 and 0
df_proc["is_global_int"] = df_proc["is_global"].astype(int)
df_proc["is_length_outlier_int"] = df_proc["is_length_outlier"].astype(int)

df_proc[["location", "location_enc", "category_clean", "category_freq", "script", "script_enc", "is_global_int"]].head()
```

**Why this encoding:** `location` and `script` have few values, so label encoding is enough. `category_clean` has thousands of unique values, so integer codes would invent a fake order; frequency encoding keeps one useful number (how common the category is) instead.


## 6. Save Processed Dataset


```python
df_proc.to_csv("../data/processed/trends_processed.csv", index=False)
print("Saved processed data to ../data/processed/trends_processed.csv")
print("Final shape:", df_proc.shape)
```
